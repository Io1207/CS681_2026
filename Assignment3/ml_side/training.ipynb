```python
train_cols = ["ArrivalRate","device_write","cpus","memory","cpuLoad","size"]
```

```python
def strip_suf(x):
    if isinstance(x, str):
        if x.endswith("mb"):
            return int(x[:-2])
        if x.endswith("m"):
            return int(x[:-1])
        if x.endswith("g"):
            return int(x[:-1]) * 1024
    return float(x)

df["device_write"] = df["device_write"].apply(strip_suf)
df["memory"] = df["memory"].apply(strip_suf) 
```

```python
df_train, df_test = train_test_split(df, test_size=0.2)
```
